## Data cleaning

##### I didn't think dropping the null values would make sense because it could be those people didn't experience any Physical health issues and/or Mental Health issues.

```python
post_pandemic_df["Physical_Health_Issues"] = post_pandemic_df["Physical_Health_Issues"].fillna("None")
post_pandemic_df["Physical_Health_Issues"] = post_pandemic_df["Physical_Health_Issues"].replace(r'^\s*$', "None", regex=True)

post_pandemic_df["Physical_Health_Issues"].value_counts()
```

```python
post_pandemic_df['Mental_Health_Status'] = post_pandemic_df['Mental_Health_Status'].fillna("None")
post_pandemic_df['Mental_Health_Status'] = post_pandemic_df['Mental_Health_Status'].replace(r'^\s*$', "None", regex=True)

post_pandemic_df['Mental_Health_Status'].value_counts()
```

## Exploding the Physical Health Issue column because is have 2 or more value in a single cell, which is against the normalization rule.

```python
#Creating a unique id for the dataset
post_pandemic_df = post_pandemic_df.reset_index(drop=True)

post_pandemic_df['Person_ID'] = post_pandemic_df.index + 1

post_pandemic_df.head()
```

```python
# 1. Pop the column out of the DataFrame
col = post_pandemic_df.pop('Person_ID')

# 2. Insert it back at index 0 (the first position)
post_pandemic_df.insert(0, 'Person_ID', col)
```

```python
# Split multiple health issues
post_pandemic_df['Physical_Health_Issues'] = post_pandemic_df['Physical_Health_Issues'].str.split(';')

post_pandemic_df.head()
```

```python
post_pandemic_df['Physical_Health_Issues'] = post_pandemic_df['Physical_Health_Issues'].apply(
    lambda x: [i.strip() for i in x] if isinstance(x, list) else x
)
post_pandemic_df.head()
```

```python
post_pandemic_df = post_pandemic_df.explode('Physical_Health_Issues')

post_pandemic_df.head()
```

## Changing the datatype for Survey_Date to date

```python
post_pandemic_df['Survey_Date'] = pd.to_datetime(post_pandemic_df['Survey_Date'])

post_pandemic_df.info()
```

## Standardizing the columns

```python
col = post_pandemic_df.columns
col
```

```python
post_pandemic_df.columns = [col.lower() for col in post_pandemic_df.columns]

post_pandemic_df.head()
```
